```python
# create array
list1 = [1,2,3]
print(list1)
```

```python
arr_list1 = np.array(list1)
print(arr_list1)  
```

```python
print(type(arr_list1))  # n-dimensional array(multi-dimensional type)
```

```python
list1 = [1,2,3]
list2 = [4,5,6]
arr_list1_2 = np.array([list1,list2])
print(arr_list1_2)  # 2d array axio
```

```python
# axis 0 -> rows (upper se neche)
# axix 1 -> columns ( baye se daye)
# 2 Rows
# 3 columns
# arr[0][0] = 1
# arr[1][2] = 6
# pahla index = row
# dusra index = column
```

```python
list1 = [1,2,3]
list2 = [4,5,6]
list3 = [7,8,9]
arr_list_3 = np.array([[list1,list2,list3]])
print(arr_list_3)
print(arr_list_3.ndim)
```

```python
a = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(a)
print(a.ndim)
```
