```python
# Statistical Functions
```

```python
a = np.array([[1,2,3],[4,5,6]])
print(a)
```

```python
print(np.sum(a))  # sum of all elements of an array
```

```python
print(np.mean(a))  # average
print(np.median(a)) # middle value
print(np.std(a)) # standard deviation (spread)
print(np.min(a)) # smallest
print(np.max(a)) # largest
```

```python
a = np.array([1,5,3])
b = np.array([4,5,6])
print(a)
print(b)
```

```python
print(a==b) # comparing each elements of an array 
print(np.array_equal(a,b)) # comparing complete array
```

```python
# copy vs view()
```

```python
x = np.array([1,2,3,4])
y = x.copy() # original array ka poora naya duplicate banata hai
z = x.view() # original array ka reference banata hai 

# x[0] = 200
# print(x) # x change hua
# print(y) # same raha

# view 
x[1] = 300
print(x)
print(z) # original array ka reference banata hai
```
