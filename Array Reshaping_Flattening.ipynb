```python
import numpy as np 
```

```python
# Reshape Array :  Array ke data ko same rakthe hue uska structure change karna
# Flatten Array : 2d and 3d array ko ek single line 1d me convert karna
```

```python
arr1 = np.array([1,2,3,4,5,6])  # shape = (1,6) - 1 row  and 6 columns
print(arr1)
```

```python
reshaped = arr1.reshape((6,1))  # shape = (6,1) - 6 rows and 1 column
print(reshaped)
```

```python
reshaped2 = arr1.reshape((2,3))
print(reshaped2)
```

```python
print(reshaped2.flatten()) 
```

```python
# Array Stacking : do ya zayada arrays ko jodna (combine karna).  
#type 2( vertical stack np.vstack((a,b))
# horizontal stack : np.hstack((a,b))
# array splitting : ek array ko chhote parts me todna
```

```python
#stacking
a = np.array([1,2,3])
b = np.array([4,5,6])
stacked = np.stack((a,b))
print(stacked)

print(np.vstack((a,b))) # vertical stacking - row wise -> output array is 2d
print(np.hstack((a,b)))  # horizontal stacking - column wise -> output array is 1d
```

```python
# splitting 
c = np.array([[1,2,3],[4,5,6]])
print(c)
```

```python
hsplit = np.hsplit(c,3) # horixontal split --> input and output array are same dimension
print(hsplit)
# for s in hsplit:
#     print(s)
```

```python
vsplit = np.vsplit(c,2) # vertical split --> input and output array are same dimension
print(vsplit)
# for s in vsplit:
#     print(s)
```
