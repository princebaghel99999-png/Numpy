```python
# Broadcasting NumPy ka ek powerful feature hai jo different shape ke arrays par arithmetic 
#operations perform karne
# deta hai without manually repeating data.
```

```python
import numpy as np
```

```python
a = np.array([1,2,3,4])
b = np.array([5])
print(a+b)
```

```python
# 1D + 2D
arr1 = np.array([[1,2,3],
                 [4,5,6]])

arr2 = np.array([10,20,30])

print(arr1 + arr2)
```

```python
# arr2 har row par apply hua:
# [1,2,3] + [10,20,30]
# [4,5,6] + [10,20,30]
```

```python
# handling with nan null value
```

```python
data1  = np.array([1,2,np.nan,4,np.inf])
print(data1)
```

```python
print(np.isnan(data1))
```

```python
print(np.nan_to_num(data1)) # null value ko 0 mai badalna
```

```python
# save and load arrays 
```

```python
arr = np.array([1,0,0,8])
print(arr)
np.save('my_array.npy',arr)
```

```python
loaded_arr = np.load('my_array.npy')
print(loaded_arr)
```

```python
# create 5x5 matrix with value from 1 to 25
```

```python
import numpy as np

arr = np.arange(1,26).reshape(5,5)

print("oringinal matrix:\n",arr)
```

```python
import numpy as np

arr = np.arange(1,26).reshape(5,5)

print("Original Matrix: \n",arr)

print("last row \n",arr[-1])

print('first column \n',arr[:,0])

print('3x3 submatrix \n',arr[1:4,1:4])
```

```python
# Broadcasting : multiply shape(3,1) with shape(3,)
```

```python
a = np.array([[1],
              [2],
              [3]])

b = np.array([10,20,30])

print(a * b)

#a =(3,1)
# b=(3,)

# broadcasting result:
# [[1*10, 1*20, 1*30],
# [2*10, 2*20, 2*30],
# [3*10, 3*20, 3*30]]
```
